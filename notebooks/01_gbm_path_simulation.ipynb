#### **What do random realisations of Geometric Brownian Motion (GBM) look like?**

```python
import sys
sys.path.append("../src")
import matplotlib.pyplot as plt
from gbm import simulate_gbm
```

The GBM stock price model is as follows:
$$
dS_t = \mu S_t\,dt + \sigma S_t\, dW_t
$$
where $S_t$ is the spot price at time $t$, $\mu$ is the drift, $\sigma$ is the volatility, and $W_t$ is standard Brownian motion.<br><br>
For a small time-step $\Delta t$, we have
$$
\Delta W_t = \sqrt{\Delta t}\,Z
$$
with $Z \sim N(0, 1)$.<br><br>
Using this formula, we naturally get the Euler-style discrete time-step approximation, which is:
$$
S_{t+\Delta t} \approx S_t + \mu S_t\Delta t + \sigma S_t\sqrt{\Delta t}\,Z
$$
However, there is also an exact solution for $S$ at time $t$, based upon its initial value, $S_0$:
$$
S_t = S_0\exp{\left[\left(\mu - \frac{1}{2} \sigma^2\right)t + \sigma\sqrt{t}\,Z\right]}
$$

We can plot several realisations of a geometric Brownian motion path, to illustrate the randomness and direction of the process with a given volatility and drift:

```python
n_paths = 10

price_paths = simulate_gbm(
    n_paths=n_paths,
    drift=0.05,
    vol=0.2,
    S0=100,
    T=1,
    dt=1/252
)

plt.figure()

plt.title(f"Example GBM paths (n = {n_paths})")
plt.xlabel("Time (t)")
plt.ylabel("Spot price (S)")

for path in price_paths:
    plt.plot(path)

plt.show()
```
