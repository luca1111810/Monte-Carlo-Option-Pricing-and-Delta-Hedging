## **Can we use Monte Carlo simulations to value Asian options?**

```python
import sys
sys.path.append("../src")
import matplotlib.pyplot as plt
from monte_carlo import MC_value_asian_option
```

The payoff of Asian options is based upon the mean spot price $\bar{S}$ over the life of the option. So:
$$
V_{Asian} = e^{-rT}\max(\bar{S} - K,\,0)
$$
<br>
We predict option prices with confidence values using our Monte Carlo method, adjusted for this difference in option payoff from European options:

```python
option_price_estimate, confidence_bounds = MC_value_asian_option(
    call=True,
    n_paths=10_000,
    S0=100,
    K=100,
    r=0.05,
    vol=0.2,
    T=1,
    alpha=0.05
)

print(f"Option price estimate: £{round(option_price_estimate, 2)}")
print(f"Confidence interval: ({round(confidence_bounds[0], 2)}, {round(confidence_bounds[1], 2)})")
```
